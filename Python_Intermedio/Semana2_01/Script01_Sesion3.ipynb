# ¿Que es una clase en Python?

Una clase es un objeto de primer orden que define una plantilla (blueprint) para crear instancias conocidas como objetos.

**Una clase en python es una fabrica de objetos**

# Fundamentos Teoricos

## Objeto como Tipo de Dato Abstracto con comportamiento

- TDA (Tipo de Dato Abstracto) es una construccion logica-matematica
    - Una representacion, que es la estructura concreta que materializa los valores del tipo
    - Una invariante de representacion
- Objeto : Es una entidad computacional en tiempo de ejecucion




    - Rol de tipo (interfaz) : Define la signatura de operaciones visibles al cliente (el protocolo del objeto). En este rol, la clase es la especificacion del TDA
    - Rol de modulo (implementacion) : Define la representeacion (los atributos) y los cuerpos de los metodos. En este rol la clase es la unidad de encapsulamiento que oculta la representacion.



## Ejemplito : `TransaccionFinanciera`

### Objetivo:
- Sea una especificacion de TDA : defina una signatura de operaciones (interfaz publica) y las propiedades que esas operaciones garantizan
- Sea una encarnacion de objeto: Cada instancia tenga identidad propia, estado privado y comportamiento seleccionable
- Sea verificable : Las propiedades se expresen como invariantes de clase y contratos de operacion (no como comentarios)

### Requerimientos Funcionales (RF)
- **RF-1 (Estados del ciclo de Vida)** : Toda transaccion debe encontrarse en uno de cuatro estados :
    - `PENDIENTE`
    - `PROCESADA`
    - `RECHAZADA`
    - `REVERSADA`

- **RF-2 (Procesamiento)** : Una transaccion en estado `PENDIENTE` debe poder procesarse, pasando a estado `PROCESADA`y registrando el evento correspondiente en su historial. Si la transaccion no esta en `PENDIENTE`, la operacion debe rechazarse por violacion de precondicion

- **RF-3 (Reversion)** : Un transaccion en estado `PROCESADA` debe poder reverzarse indicando un motivo, pasando a estado `REVERSADA` y registrando el evento correspondiente en su historial. Si la transaccion no esta en `PROCESADA`, entonces la operacion debe rechazarse por violacion de precondicion.

- **RF-4 (Validacion Previa)** : Debe existir una operacion de consulta que, sin modificar el estado, evalue las reglas de negocio aplicables y devuelva un resultado de validacion

- **RF-5 (Exportacion)** : Debe existir una operacion de consulta que, sin modificar el estado, produzca una representacion de la transaccion en un formato solicitado

### Requerimientos de Diseño  del TDA (RD)
1. Identididad
    - **RD-1 (Identidad Referencial)**
    - **RD-2 (Identidad de dominio persistente)**
    - **RD-3 (Distincion Explicita)**
2. Estado (Representacion Privada)
    - **RD-4 (Ocultamiento de la representacion)**
    - **RD-5 (Atributos de estado)** : Es estado debe de constar exactamente de:
        - `_monto`
        - `_fecha_valor`
        - `_estado`
        - `_historial`
    - **RD-6 (Invariante de Representacion)**
3. Comportamiento (interfaz Publica)
    - **RD-7 (Separacion comando-consulta)**
        - Consultas
        - Comandos
    - **RD-8 (Contratos de los comandos)**
4. Invariantes de Clase
    - **RD-9 (Invariante de Clase)**
5. Propiedades de Historia
    - **RD-10 (Maquina de Estados)**
    - **RD-11 (Historial append-only)**

### Restricciones Tecnicas (RT)
- **RT-1** : El lenguaje debe soportar tipado estatico y aserciones ejecutables (precondiciones, postcondiciones, invariantes) o , en su defecto, un mecanismo equivalente verificable en tiempo de ejecucion.
- **RT-2** : La representacion del estado debe ser independiente de la interfaz
- **RT-3** : La identidad de dominio debe generarse con un generador `UUID v4`
- **RT-4**:El diseño debe documentar explicitamente, para cada propiedad, si es invariante de clase o propiedad de historia , y justificar la clasificacion.

### Entregables
- Diagrama de la clase
- Especificacion contractual de la clase
- Documentacion de las propiedades de historia
- Codigo de clase en un lenguaje que satisfaga las RT

Clase (TDA) : `TransaccionFinanciera`
- IDENTIDAD
    - Referencial (POO) : Referencia en memoria
    - de dominio (persistente)
- ESTADO
    - `_monto` : DECIMAL

    - `_historial`:Lista Inmutable de eventos (estructura persistente)
- COMPORTAMIENTO (Interfaz Publica)
    - CONSULTAS : validar_reglas_negocio(), exportar_a_formato(formato)
    - COMANDOS :
        - procesar()
        - reversar(motivo)
- INVARIANTE DE CLASE
    - monto>0
    - estado este en {PENDIENTE, PROCESADA, RECHAZADA , REVERSADA}
    - historial siempre sea inmutable
- PROPIEDADES DE LA HISTORIA
    - Toda secuencia de estados es un camino valido de la maquina de estados
    - historial es append-only

````python
# Clase de Ejemplo : Usuario
class Usuario:
    """
    Al no poner ninguna clase padre en Usuario, la clase usuario es clase hija directa de
    la clase object (la clase padre de todo en python)
    """

    #
    # Aca puedo poder atributos de clase
    #

    def __init__(self, nombre, email, historial_compras, metodo_autenticacion, fecha_registro):
        """
        Documentacion de la naturaleza de los elementos de cada instancia de clase
        """

        # Atributos de instancia
        self.nombre = nombre
        self.email = email
        self.historial_compras = historial_compras
        self.metodo_autenticacion = metodo_autenticacion
        self.fecha_registro = "Resultado de una consulta a una BD del sistema"
        self.carrito = []
        self.activo = True

    # Luego de semanas de debate se decidi que patron de diseño utilizar
    def agregar_ar_carrito(self, producto):
        """

        """
        if self.activo:
            self.carrito.append(producto)
        else:
            raise ValueError("Usuario : Inacntivo")
````
