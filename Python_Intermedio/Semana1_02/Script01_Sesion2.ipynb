## Flujo General de una Excepcion

1. Se ejecuta codigo dentro de un bloque `try`
2. Ocurre un error o se usa explicitamente el comando `raise`
3. Python crea o instancia un objeto de una excepcion
4. Se busca un bloque `except` comptatible
    * Primero en el mismo bloque
    * Luego en funciones superiores (propagacion)
    * Si no se captura, el programa debe terminar en `traceback`
5. Si no ocurre excepcion -> Se ejecuta el bloque `else`
6. Independientemente del resultado -> Se ejecuta el bloque `finally`

# Definicion de Excepciones Personalizadas

1. Heredar siempre de `Exception` (No de `BaseException`)
2. Utilizar el sufijo `Error` (Por convencion)
3. Definir una jerarquia base para nuestro dominio de aplicacion o modulo.
4. Añadir atributos y mensajes informativos
    - Definir `__init__` para recibir informacion contextual y almacenarla como atributos
    - Sobreescribir : `__str__`
5. Mantener la excepcion ligera (evitar logica compleja)
6. Documentar las excepciones personalizadas
7. Usar herencia multiple solo si es necesario (con cuidado)
8. No abusar de las excepciones personalizadas para casos cubiertos por el estandar (modulo : `builtins`)
9. Lanzar siempre con `raise` y proporcionar contexto con `from`
10. Usar clases vacias (solo `pass`) cuando solo importe el tipo
11. Considerar usar `__slots__` para ahorrar memoria (opcional)
12. Exportar las excepciones en `__init__` del modulo

```python
# El primer item no dice que siempre debemos heredar de Excepcion
class MiExcepcion(Exception):
    """
    Documentacion mediante docstring
    Descripcion del error
    """

    pass
```

```python
# Podemos lanzar o ejecutar nuestra excepcion personalizada
raise MiExcepcion("Mensaje Descriptivo")
```

```python
# Ejemplito
# Definamos una excepcion personalizada para validad una edad
# Si la edad es negativa (o si es mayor que 115) se lanza una excepcion

class EdadInvalidaError(Exception):
    """
    Se lanza/instancia cuando la edad es menor que cero o mayor o
    igual que 115
    LO cual proviene de suposiciones legales o por contrato con el cliente (Area Usuaria)
    """
    pass

def validad_edad(edad):
    if (edad<0) or (edad>=115):
        raise EdadInvalidaError(f"Edad Invalidad : {edad} : Debe encontrarse en [0,115]")
    return f"Edad Valida : {edad}"
```

```python
validad_edad(12)
```

```python
validad_edad(-5)
```

```python
# Podemos utilizar try-except
try:
    print(validad_edad(0))
    print(validad_edad(25))
    print(validad_edad(-2))
    print(validad_edad(-36))
    print(validad_edad(45))
    print(validad_edad(98))
    print(validad_edad(116))
except EdadInvalidaError as e :
    print(f"Error Capturado : {e}")
```

```python
# Modifiquemos la logica de las pruebas anteriores

# Puedo llevar la cuenta de las veces que se utilizo mi excepcion
num_llamadas = 0

for edad in [0,25,-2, -36, 45,98,116]:
    try:
        print(validad_edad(edad))
    except EdadInvalidaError as e:
        num_llamadas = num_llamadas +1
        print(f"Error capturado : {e}")

# Mostremos reporte
print(f"Numero de llamadas : {num_llamadas}")
```

```python
# Ejemplito : Validacion un email
# Definir una excepcion
# Para validar que un email (str) contenga : "@" y "."

class EmailInvalidoError(Exception):
    """
    Email no cumple con un formato basico
    """
    pass

def validar_email(email):
    if "@" not in email or "." not in email:
        raise EmailInvalidoError(f"Email Invalido : {email}")
    return "Email Valido"
```

```python
# Ejecutemos algunas pruebas
lista_correos_validar = [
    "azamudio@pit788.uni.edu.pe",
    "usuario_arroba_linux.pe",
    "123456789_mit.edu.pe",
    "wcastillo@uni.pe"
]

for email in lista_correos_validar:
    try :
        print(validar_email(email))
    except EmailInvalidoError as e:
        print(f"Error : {e}")
```

```python
# Ejemplito : Validacion de coordenadas geograficas
# Definamos una excepcion : CoordenadaInvalidaError
# Valida que la latitud este en [-90,90] y la longitud en [-180,180]

class CoordenadaInvalidaError(Exception):
    """
    PIT788 : Coordenada fuera de rango valido
    """
    pass

def validar_coordenada(coordenadas):
    """
    coordenadas[0] : lat
    coordenadas[1] : lon
    """
    lat = coordenadas[0]
    lon = coordenadas[1]
    if not (-90 <= lat <= 90):
        # Mediante alguna logica de negocio
        raise CoordenadaInvalidaError(f"Latitud Invalida : {lat}")
    if not (-180 <= lon <= 180):
        # Mediante alguna logica de negocio
        raise CoordenadaInvalidaError(f"Longitud Invalida : {lon}")
    # Luego de alguna logica de negocio
    return "Coordenadas Validas"
```

```python
# Ejecutemos algunas pruebas
lista_gps = [(-407.128,-74.0060), (100,-47), (-11, 19)]

for gps in lista_gps:
    try:
        print(validar_coordenada(gps))
    except CoordenadaInvalidaError as e :
        print("Ejecutar: Rutina Mantenimiento")
        print(f"Error : {e}")
```

# ¿Que es una clase en Python?

Una clase es un objetoo de primer orden que define una plantilla (blueprint) para crear instancias conocidas como objetos

|Programacion Estructurada | programacion orientada a objetos |
|---------------------------|------------------------|
|Funcion|Objeto/Clase|
| Variables globales o pasadas por referencia | Encapsulado dentro del objeto|
| Secuencial, Explicito, Jerarquico| Basado en mensajes entre objetos |
| Las funciones dependen de estructuras de datos compartidas | Los objetos depende de interfaces/contratos mas no de implementaciones|
|Agrupacion por proximidad funcional | Agrupacion por responsabilidad unica|
